# Loss functions for clasifications
Have to choose the appropiate loss function, depending on the type of problem (binary versus multiclass) and type of output (logit versus probabilities).

### Binary cross-entropy
Is the correct loss function for a binary-classification (with a single output unit)

### Categorical cross-entropy
Is the appropiate loss function for multiclass-classification. If the ground thuth are one-hot-encoded format for example [0,1,0].

### Spare categeorical cross-entropy 
Is also for multicalss-classification, but if the ground thruth are not one-hot-encoded format, but provided as integear labels for example y=2

Each of these three loss functions can also recive the prediticions in form of logit or class-membership probabilities

Computing the crosss-entropy loss by providing logits and not the class-membership probabilities, is usually prefered due to numerical stability reasons. If we provide logits as input to the loss functions and set 'from_logits=True' TensorFlow uses a more efficient calculations.

![Loss functions](loss_functions.png)

```python
# code to show how to use the different loss functions with two different formats, 
# where either logits or class-memberships probabilities are given as input to the functions
import tensorflow_datasets as tfds
import tensorflow as tf
```

```python
#binary cross-entropy
bce_probas = tf.keras.losses.BinaryCrossentropy(from_logits=False)
bce_logits = tf.keras.losses.BinaryCrossentropy(from_logits=True)

logits = tf.constant([0.8])
probas = tf.keras.activations.sigmoid(logits)

# wasn't in the book. But without it got error, complaining on ValueError. Solution making both in tf.tensors
y_true = tf.constant([1.0])

tf.print('BCE (w probas): {:.4f}'.format(bce_probas(y_true=y_true, y_pred=probas)),
         '(w logits): {:.4f}'.format(bce_logits(y_true=y_true, y_pred=logits)))
```

```python
# cateogircal cross-entropy
cce_probas = tf.keras.losses.CategoricalCrossentropy(from_logits=False)
cce_logits = tf.keras.losses.CategoricalCrossentropy(from_logits=True)

logits = tf.constant([1.5, 0.8, 2.1])
probas = tf.keras.activations.softmax(logits)

y_true = tf.constant([0, 0, 1])

tf.print('CCE (w probas): {:.4f}'.format(cce_probas(y_true=y_true, y_pred=probas)),
         '(w logits): {:.4f}'.format(cce_logits(y_true=y_true, y_pred=logits)))
```

```python
# spare categorical cross-entropy
sp_cce_probas = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False)
sp_cce_logits = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)

y_true = tf.constant([2])

tf.print('Sparse CCE (w probas): {:.4f}'.format(sp_cce_probas(y_true=y_true, y_pred=probas)),
         '(w logits): {:.4f}'.format(sp_cce_logits(y_true=y_true, y_pred=logits)))
```
