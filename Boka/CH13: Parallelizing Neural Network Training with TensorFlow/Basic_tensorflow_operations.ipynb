# Creating Tensors in TensorfFlow
A few diffeerent ways to create tensors

```python
import tensorflow as tf
```

```python
# create tensor from a list or Numpyarray using tf.convert_to_tensor
import numpy as np

np.set_printoptions(precision=3)

a = np.array([1,2,3], dtype=np.int32)
b = [4,5,6]

t_a = tf.convert_to_tensor(a)
t_b = tf.convert_to_tensor(b)

print(t_a)
print(t_b)
```

```python
# Similar to NumPy arrays, we can further see the properties
t_ones = tf.ones((2,3))
t_ones.shape
```

```python
# To get access to tha values that a tensor refers to we can, call the numpy() method on a tensor
t_ones.numpy()
```

```python
# Creating a tensor of constant values can be done as follows
const_tensor = tf.constant([1.2, 5, np.pi], dtype=tf.float32)
print(const_tensor)
```

# Manipulating the data type and shape of a tensor
Manipulate tensors data types and shapes via several TensorFlow funcitons that cast, reshape, transpose, and sequence

```python
t_a_new = tf.cast(t_a, tf.int64)
print(t_a_new.dtype)
```

```python
#Transposing a tensor
t = tf.random.uniform(shape=(3, 5))
t_tr = tf.transpose(t)
print(t.shape, '--->', t_tr.shape)
```

```python
#Reshaping a tensor (f.eks. from 1D vector to a 2D array)
t = tf.zeros((30,))
t_reshape = tf.reshape(t, shape=(5,6))
print(t_reshape.shape)
```

```python
#Removing the unnecessary dimensions (dimensions that have size 1, which are not needed)
t = tf.zeros((1, 2, 1, 4, 1))
t_sqz = tf.squeeze(t, axis=(2,4))
print(t.shape, '--->', t_sqz.shape)
```

# Applying mathematical operations to tensors
Widley used linear algebra operations, element-wise product, matrix multiplication, and computing norm of a tensor

```python
#Initiate two random tensors, one with uniform distirbution in the range [-1, 1) and the other with standard normal distribution
tf.random.set_seed(1)

t1 = tf.random.uniform(shape=(5,2), minval=-1.0, maxval=1.0)
t2 = tf.random.normal(shape=(5,2), mean=0.0, stddev=1.0)
print('t1:', t1)
print('t2:', t2)
```

```python
#Compute element-wise product of t1 and t2
t3 = tf.multiply(t1,t2).numpy()
print(t3)
```

```python
#To compute the mean,sum and the standard deviation along a ceratin axis (or axes), we can use
# tf.math.reduce_mean(), tf.math.reduce_sum(), and tf.math.reduce.std()

#Mean of each column in t1 can be computed:
t4 = tf.math.reduce_mean(t1, axis=0)
print(t4)
```

```python
# Matrix product between t1 and t2 (t1 x t2.T) can be calculated using tf.linalg.matmul()
t5 = tf.linalg.matmul(t1, t2, transpose_b=True)
print(t5.numpy())
```

```python
# Or on another hand, computing t1 x t2.T is prefromed by transposing t1, resulting in an array 2x2
t6 = tf.linalg.matmul(t1, t2, transpose_a=True)
print(t6.numpy())
```

```python
# The norm function is usefull for computing the L^P norm of a tensor
# F.eks. we can calulcate the L^2 norm of t1 as follows
norm_t1 = tf.norm(t1, ord=2, axis=1).numpy()
print(norm_t1)

# verify with numpy that it is correct
print(np.sqrt(np.sum(np.square(t1), axis=1)))
```
